# Python 練習 チューター（Gemini 自動解説）

**使い方**:
1. セル① で API キーを設定（1回だけ）
2. セル② に今日の問題を貼る
3. セル③ に自分のコードを書く
4. セル④ を実行 → Gemini がチェックしてくれる

**API キーの取得**: `aistudio.google.com` → 「Get API key」→ 無料で作れます

```python
# ==============================
# セル① セットアップ（最初に1回だけ実行）
# ==============================
!pip install -q google-generativeai

import google.generativeai as genai
import io, sys

# ↓ ここに自分の API キーを貼る
API_KEY = "AIza...ここにキーを貼る..."

genai.configure(api_key=API_KEY)
model = genai.GenerativeModel("gemini-1.5-flash")
print("準備完了！")
```

```python
# ==============================
# セル② 今日の問題を貼る
# （VS Code の daily_problems/ から問題文をここにコピー）
# ==============================
PROBLEM = """
【ここに問題を貼る】
例:
5天体の名前と距離を表示する関数 show_target(name, distance) を書いてください。
期待出力: "Draco: 距離 76 kpc"
"""

print(PROBLEM)
```

```python
# ==============================
# セル③ ここに自分のコードを書く！
# ==============================

# ↓ ここから書く

def show_target(name, distance):
    print(f"{name}: 距離 {distance} kpc")

show_target("Draco", 76)
show_target("Coma Berenices", 44)
```

```python
# ==============================
# セル④ Gemini にチェックしてもらう（実行するだけ！）
# ==============================
import inspect, traceback

# セル③ のコードをテキストとして取得
import IPython
history = IPython.get_ipython().history_manager
# 直前のセルのソースコードを取得
cells = list(history.get_range())
user_code = cells[-2][2] if len(cells) >= 2 else "（コード取得失敗）"

# 実際に実行して出力を取得
stdout_capture = io.StringIO()
sys.stdout = stdout_capture
try:
    exec(user_code, {})
    actual_output = stdout_capture.getvalue()
    exec_error = None
except Exception as e:
    actual_output = ""
    exec_error = traceback.format_exc()
finally:
    sys.stdout = sys.__stdout__

# Gemini に送るプロンプト
prompt = f"""あなたは Python 初心者（Level 0〜5）のやさしい家庭教師です。
日本語で、3〜5行以内で簡潔にフィードバックしてください。
褒めるときは短く、間違いは「ここが違うよ」と具体的に。
難しい用語は使わないで。

【今日の問題】
{PROBLEM}

【生徒のコード】
{user_code}

【実行結果】
{actual_output if not exec_error else '実行エラー:\n' + exec_error}

フィードバックをお願いします。"""

response = model.generate_content(prompt)
print("=" * 40)
print("Gemini のフィードバック:")
print("=" * 40)
print(response.text)
```
